```python
#DATA CLEANING

import pandas as pd

#check missing values in each column
df = pd.read_csv("healthcare_dataset.csv")
print("missing values in each columns")
print(df.isnull().sum())

#calculate the missing values percentages

print("\n missing values percentage:")
print((df.isnull().sum()/len(df))*100)

#only c0lumn contains missing values

missing = df.isnull().sum()

print([missing>0])

```

```python
#2)check duplicate records


import pandas as pd

df = pd.read_csv("healthcare_dataset.csv")

#count duplicate rows

print("total duplicate rows:",df.duplicated().sum())

#displaying duplicate records

duplicates = df[df.duplicated(keep=False)]
display(duplicates.head(5))

#removing exact duplicate rows

df = df.drop_duplicates()

print("dataset after removing duplicated rows:",df.shape)
```

```python
#3)correcting data types

import pandas as pd
#displaying each column datatype
df = pd.read_csv("healthcare_dataset.csv")
print(df.dtypes)

#convert date column to date time


df["Date of Admission"] = pd.to_datetime(
    df["Date of Admission"], errors="coerce"
)

df["Discharge Date"] = pd.to_datetime(
    df["Discharge Date"], errors="coerce"
)
#check the updated datatype

print(df[["Date of Admission","Discharge Date"]].dtypes)

#how many date are missing

print("missing Admission Date:",df["Date of Admission"].isnull().sum())
print("missing Discharge Date:",df["Discharge Date"].isnull().sum())
```
